#**Step 2: Data Cleaning**
The dataset was checked for missing values, duplicate records, and data quality issues before analysis.

```python
df["cast"].fillna("Unknown", inplace=True)
```

```python
df["country"].fillna("Unknown", inplace=True)
```

```python
df["date_added"].fillna("Unknown", inplace=True)
```

```python
df["rating"].fillna("Unknown", inplace=True)
```

```python
df["duration"].fillna("Unknown", inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.duplicated().sum()
```

### Data Cleaning Decisions

1. Missing director names were replaced with "Unknown".
2. Missing cast information was replaced with "Unknown".
3. Missing country values were replaced with "Unknown".
4. Missing ratings were replaced with "Not Rated".
5. Missing duration values were replaced with "Unknown".
6. Missing dates were filled using the most frequent date.
7. Duplicate records were removed.

#5. Exploratory Data Analysis
The internship requires 5 analysis questions.

##Question 1
What is the distribution of Movies and TV Shows?

```python
df["type"].value_counts()
```

```python
df["type"].value_counts().plot(kind="bar")
plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
```

##Question 2
Which countries produce the most Netflix content?

```python
df["country"].value_counts().head(10)
```

```python
df["country"].value_counts().head(10).plot(kind="bar")
plt.title('Countries with the Most Netflix Content')
plt.xlabel('Country')
plt.ylabel('Count')
```

##Question 3
Which ratings are most common?

```python
df["rating"].value_counts()
```

```python
df["rating"].value_counts().plot(kind="bar")
plt.title('Common Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
```

##Question 4
Which release years have the highest content count?

```python
df["release_year"].value_counts()
```

```python
df["release_year"].value_counts().head(10).plot(kind="bar")
plt.title('Release Years with the Highest Content Count')
plt.xlabel('Release Year')
plt.ylabel('Count')
```

##Question 5
What are the most common genres?

```python
df["listed_in"].value_counts()
```

```python
df["listed_in"].value_counts().head(10).plot(kind="bar")
plt.title('Most Common Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
```

#**6. Visualizations**
##Bar Chart

Movies vs TV Shows

```python
df["type"].value_counts().plot(kind="bar")
plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
```

##Histogram

Release year distribution

```python
df["release_year"].hist()
plt.title('Release Year Distribution')
plt.xlabel('Release Year')
plt.ylabel('Count')
```

##Pie Chart

Content type distribution

```python
df["type"].value_counts().plot(kind="pie", autopct='%1.1f%%')
plt.title('Movies vs TV Shows')
```

##Scatter Plot

Release year vs duration

```python
sns.scatterplot(
x="release_year",
y="duration",
data=df
)
plt.title('Release Year vs. Duration')
plt.xlabel('Release Year')
plt.ylabel('Duration')
```

```python
country_type = pd.crosstab(
    df['country'],
    df['type']
).sort_values(
    by='Movie',
    ascending=False
).head(10)

plt.figure(figsize=(10,6))
sns.heatmap(country_type,
            annot=True,
            cmap='YlGnBu')

plt.title("Top 10 Countries vs Content Type")
plt.show()
```
